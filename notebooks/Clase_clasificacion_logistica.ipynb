## Importar las librerias a usar.

```python
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

### Carga de Datasets

```python
X_train = pd.read_csv('../clasificacion/X_train.csv')
X_test = pd.read_csv( '../clasificacion/X_test.csv')
y_train = pd.read_csv('../clasificacion/y_train.csv').squeeze()
y_test = pd.read_csv('../clasificacion/y_test.csv').squeeze()
```

```python
print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)

print("\nColumnas de X_train:")
print(list(X_train.columns))
```

```python
#Modelo

modelo = LogisticRegression(max_iter=1000)
modelo.fit(X_train, y_train)

print("Modelo entrenado con", X_train.shape[0], "Pasajeros de entrenamiento")

#Prediccion y Probabilidades
y_pred = modelo.predict(X_test)
y_proba = modelo.predict_proba(X_test)

print("Predicciones (primeras 5):", y_pred[:5])
print()
print("Probabilidades (primeras 5): - [Sobrevivio o no Sobrevivio]")
print(y_proba[:5].round(3))
print()
print("Valores reales (primeras 5)", y_test.values[:5])
```

```python
#Metricas
accuracy = accuracy_score(y_test, y_pred)

#:.4f esto se denomina un especificador de formato para decirle al codigo que tantos decimales presentar.
print(f'Accuracy: {accuracy:.4f} ({accuracy * 100:.1f}%)')
```
